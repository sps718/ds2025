## Section 1: Linear Algebra Refresher & Bottom-Up Concepts

In this section we review basic vector operations. As described in the lab plan ( ), we will create vectors representing basketball players’ statistics and compute their magnitudes, dot product, and cosine similarity.

```python
import numpy as np

# Define vectors for two basketball players' statistics
# For example: [points, rebounds, assists]
player1 = np.array([25, 5, 7])
player2 = np.array([30, 4, 9])
```

The code above imports numpy and creates two NumPy arrays `player1` and `player2`, each representing statistics (points, rebounds, assists) for two basketball players. This sets up the data for subsequent vector operations.

```python
# Calculate the dot product of the two vectors
dot_product = np.dot(player1, player2)

# Compute the Euclidean norms (magnitudes) of each vector
norm_player1 = np.linalg.norm(player1)
norm_player2 = np.linalg.norm(player2)

# Calculate cosine similarity between the two players
cosine_similarity = dot_product / (norm_player1 * norm_player2)

print('Dot Product:', dot_product)
print('Norm of Player 1:', norm_player1)
print('Norm of Player 2:', norm_player2)
print('Cosine Similarity:', cosine_similarity)
```

This cell performs key vector operations:
- It computes the dot product of `player1` and `player2` using `np.dot`.
- It calculates the Euclidean norm (or magnitude) of each vector using `np.linalg.norm`.
- It then computes the cosine similarity, a measure of how similar the two vectors are, by dividing the dot product by the product of the norms.

The printed outputs will display the dot product, the magnitudes of both vectors, and the cosine similarity value.

## Section 2: Data Standardization and Covariance Matrix

This section focuses on centering the data through standardization and computing the covariance matrix. As outlined in the lab plan (fileciteturn0file1), this step is crucial before applying PCA.

```python
import pandas as pd

# Load the basketball dataset into a DataFrame
# The file 'basketball_stats.csv' should be placed in the working directory
df = pd.read_csv('basketball_stats.csv')
df.head()
```

This code uses pandas to load a basketball statistics dataset from a CSV file named `basketball_stats.csv`. The `df.head()` function displays the first few rows to confirm that the data has been loaded successfully.

```python
# Standardize the numeric columns of the dataset
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns
standardized_df = df.copy()
standardized_df[numeric_cols] = (df[numeric_cols] - df[numeric_cols].mean()) / df[numeric_cols].std()
standardized_df.head()
```

The above cell standardizes the numeric features in the dataset. It first selects all numerical columns, and then for each numeric column, it subtracts the mean and divides by the standard deviation. This re-centers the data around zero and scales it to unit variance. The result is displayed using `head()`.

```python
# Compute the covariance matrix of the standardized data
cov_matrix = standardized_df[numeric_cols].cov()
cov_matrix
```

This code calculates the covariance matrix for the standardized numerical features using the pandas `cov()` function. The covariance matrix shows how different features vary together, which is a key preparatory step for PCA.

## Section 3: Manual Implementation of PCA Using Numpy

In this section we manually implement PCA using numpy. We will compute the eigenvalues and eigenvectors of the covariance matrix, sort them, and project the data onto the principal component. This follows the conceptual approach described in the lab plan (fileciteturn0file2).

```python
# Compute eigenvalues and eigenvectors of the covariance matrix
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix.values)

print('Eigenvalues:', eigenvalues)
```

This cell uses numpy's `linalg.eig` function to perform eigen decomposition on the covariance matrix. The resulting eigenvalues indicate the amount of variance captured by each principal component, while the eigenvectors provide the direction of these components.

```python
# Sort the eigenvalues and corresponding eigenvectors in descending order
sorted_indices = np.argsort(eigenvalues)[::-1]
sorted_eigenvalues = eigenvalues[sorted_indices]
sorted_eigenvectors = eigenvectors[:, sorted_indices]

print('Sorted Eigenvalues:', sorted_eigenvalues)
```

Here we sort the eigenvalues in descending order using `np.argsort` and reorder the columns of the eigenvector matrix accordingly. This ensures that the first eigenvector corresponds to the direction of maximum variance.

```python
# Project the standardized data onto the first principal component
pc1 = np.dot(standardized_df[numeric_cols].values, sorted_eigenvectors[:, 0])

print('Projection onto first principal component (first 5 values):', pc1[:5])
```

This cell projects the standardized data onto the first principal component by taking the dot product of the data matrix with the eigenvector associated with the largest eigenvalue. The output shows the first five projected values, illustrating how the original multidimensional data is reduced onto a single axis representing maximum variance.

## Section 4: PCA with Scikit-learn and Data Visualization

In this section we transition to using scikit-learn's PCA for dimensionality reduction and visualize the results. According to the lab plan (fileciteturn0file1), this helps in comparing the manual approach with an automated, library-driven method.

```python
from sklearn.decomposition import PCA

# Initialize PCA to reduce data to 2 components for visualization
pca = PCA(n_components=2)
principal_components = pca.fit_transform(standardized_df[numeric_cols])

print('Explained variance ratio:', pca.explained_variance_ratio_)
```

This cell uses scikit-learn's PCA to reduce the standardized data to two principal components. The `fit_transform` method fits the PCA model and applies the dimensionality reduction. The printed explained variance ratio helps understand how much variance each component captures.

```python
import matplotlib.pyplot as plt

# Create a scatter plot of the two PCA components
plt.figure(figsize=(8,6))
plt.scatter(principal_components[:, 0], principal_components[:, 1], alpha=0.7, edgecolor='b')
plt.title('PCA of Basketball Data')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.grid(True)
plt.show()
```

This block visualizes the PCA results in a 2D scatter plot. Using matplotlib, it plots the first principal component on the X-axis and the second on the Y-axis. The plot is styled with grid lines, labels, and a title to help in interpreting the distribution of data in the reduced dimensional space.

## Section 5: Exploring Alternative Dimensionality Reduction Techniques

In the final section we explore another popular technique, t-SNE, for dimensionality reduction. As specified in the lab plan ( ), t-SNE is useful for visualizing high-dimensional data in a non-linear way.

```python
from sklearn.manifold import TSNE

# Initialize t-SNE to reduce data to 2 dimensions
tsne = TSNE(n_components=2, random_state=42)
tsne_results = tsne.fit_transform(standardized_df[numeric_cols])

print('t-SNE result shape:', tsne_results.shape)
```

This code creates a t-SNE instance from scikit-learn to reduce the standardized basketball data to two dimensions. By setting a fixed `random_state`, the results become reproducible. The shape of `tsne_results` confirms the dimensionality reduction.

```python
# Visualize the t-SNE transformation using a scatter plot
plt.figure(figsize=(8,6))
plt.scatter(tsne_results[:, 0], tsne_results[:, 1], alpha=0.7, edgecolor='r')
plt.title('t-SNE of Basketball Data')
plt.xlabel('t-SNE Component 1')
plt.ylabel('t-SNE Component 2')
plt.grid(True)
plt.show()
```

This cell visualizes the output of t-SNE as a 2D scatter plot. The plot is configured with labels, a title, and a grid to help reveal any clustering or patterns in the data that might be better captured by t-SNE compared to PCA.
