## Conceptual Walkthrough of K-Means Clustering

In this section we build foundational components for the K-Means algorithm starting with distance computation and centroid calculation.

```python
def squared_euclidean_distance(a, b):
    """Compute and return the squared Euclidean distance between two points a and b."""
    return np.sum((a - b) ** 2)
```

The function defined above computes the squared Euclidean distance between two points. It subtracts one point from the other, squares the differences, and sums them. This metric is useful for determining how similar or different two points are in Euclidean space.

Challenge: Re-implement this function using `np.linalg.norm` (with the result squared) to perform the same calculation.

```python
# Compute the centroid of a set of points
points = np.array([[1, 2], [2, 3], [3, 4]])
centroid = points.mean(axis=0)
print('Centroid:', centroid)
```

This code calculates the centroid (the average position) of three data points. The centroid is determined by averaging each coordinate separately.

## Hands-On Algorithm Exploration Using a Toy Dataset

Now we will simulate a simple version of the K-Means algorithm on a small toy dataset. You will see how initial centroids, cluster assignments, and centroid updates work together.

```python
# Define a toy dataset with six 2D points
data = np.array([
    [1, 2],
    [1.5, 1.8],
    [5, 8],
    [8, 8],
    [1, 0.6],
    [9, 11]
])
print('Toy dataset:\n', data)
```

The code above creates a toy dataset consisting of six two-dimensional points. This dataset will serve as the input for our clustering simulation.

Challenge: Experiment with altering the dataset by adding or removing points and observe how it might impact subsequent clustering.

```python
# Initialize centroids by selecting two random points from the dataset
np.random.seed(42)  # Ensures reproducibility
indices = np.random.choice(len(data), 2, replace=False)
centroids = data[indices]
print('Initial centroids:\n', centroids)
```

This code randomly selects two points from the toy dataset to serve as initial centroids. A fixed random seed guarantees that the same centroids are chosen on each run.

Challenge: Modify the random seed or change the number of centroids selected to explore different initial conditions.

```python
def assign_clusters(data, centroids):
    """Assign each data point to the closest centroid using squared Euclidean distance."""
    clusters = []
    for point in data:
        distances = [squared_euclidean_distance(point, c) for c in centroids]
        clusters.append(np.argmin(distances))
    return np.array(clusters)
```

The function `assign_clusters` iterates through each data point, computes its squared Euclidean distance to each centroid, and assigns the point to the cluster with the minimum distance.

Challenge: Modify this function to use a different distance metric, such as Manhattan distance.

```python
def update_centroids(data, clusters, K):
    """Calculate new centroids as the mean of data points in each cluster."""
    new_centroids = []
    for k in range(K):
        points = data[clusters == k]
        new_centroids.append(points.mean(axis=0) if len(points) > 0 else np.array([0, 0]))
    return np.array(new_centroids)
```

The `update_centroids` function computes the new centroid for each cluster by taking the mean of all data points assigned to that cluster. It also handles the case where a cluster might have no points.

Challenge: Experiment with how you handle empty clusters (for example, by reinitializing centroids differently).

```python
# One iteration: assign clusters and update centroids
clusters = assign_clusters(data, centroids)
new_centroids = update_centroids(data, clusters, K=2)
print('Cluster assignments:', clusters)
print('Updated centroids:\n', new_centroids)
```

This code executes a single iteration of K-Means, where data points are first assigned to the nearest centroid and then new centroids are calculated based on these assignments.

Challenge: Integrate these steps inside a loop to run multiple iterations until convergence, or try varying the number of clusters.

## Determining the Optimal Number of Clusters

In this section we explore how to evaluate clustering performance. The elbow method assists in selecting the optimal number of clusters by analyzing the inertia (sum of squared distances).

```python
from sklearn.cluster import KMeans

inertias = []
ks = range(1, 5)
for k in ks:
    kmeans = KMeans(n_clusters=k, random_state=42).fit(data)
    inertias.append(kmeans.inertia_)
    print('K:', k, 'Inertia:', kmeans.inertia_)
```

This block uses Scikit-Learn's KMeans to compute and print the inertia for cluster counts ranging from 1 to 4. Observing the inertia values helps in determining the 'elbow' point beyond which additional clusters do not significantly reduce the error.

Challenge: Extend the range of k (for example, 1 to 6) to see how the inertia values evolve.

```python
plt.figure(figsize=(6, 4))
plt.plot(list(ks), inertias, marker='o')
plt.title('Elbow Method for Optimal k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.show()
```

The plot generated above shows the relationship between the number of clusters and the corresponding inertia. An 'elbow' in this plot suggests a good choice for k.

## Application to a Realistic Dataset: Music Features

Here we apply the K-Means algorithm to a realistic dataset containing music track features. This exercise simulates a real-world scenario where clustering can aid in tasks like recommendation system development.

```python
import pandas as pd

# Load the music features dataset (ensure 'music_features.csv' is in the working directory)
df = pd.read_csv('music_features.csv')
print('First five rows of the music features dataset:')
print(df.head())
```

The code above reads the music features dataset from a CSV file and prints the first five rows so you can inspect its structure.

Challenge: Explore the dataset further by examining metadata like column data types and summary statistics.

```python
# Select numeric features assumed to represent music track attributes
df_features = df.select_dtypes(include=[np.number])

# Apply K-Means clustering with k=3 on the selected features
kmeans_music = KMeans(n_clusters=3, random_state=42).fit(df_features)
print('Cluster centers for music features:')
print(kmeans_music.cluster_centers_)
```

This block selects only the numeric columns from the dataset and applies K-Means clustering with three clusters. The resulting cluster centers are printed for review.

Challenge: Change the number of clusters (e.g., set k=4) and analyze how the cluster centers differ.

```python
# Evaluate clustering quality using the elbow method on the music dataset
inertias_music = []
ks_music = range(1, 6)
for k in ks_music:
    km = KMeans(n_clusters=k, random_state=42).fit(df_features)
    inertias_music.append(km.inertia_)
    print('K:', k, 'Inertia:', km.inertia_)
```

This code computes the inertia for a range of cluster counts on the music dataset and prints the results. The output can help identify the optimal number of clusters using the elbow method.
